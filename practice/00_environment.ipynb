## データを読み込む

次のセルは CSV の場所を探し，表を読みます。ノートブックと CSV を同じフォルダーに置けば動きます。教材のリポジトリから開いた場合にも対応しています。

```python
from pathlib import Path

filename = "nyc-flights-2024-01.csv"
path = Path(filename)
if not path.exists():
    path = Path("data") / filename
if not path.exists():
    path = Path("../data") / filename

if not path.is_file():
    path = "https://raw.githubusercontent.com/sohei-takagi/python-ds-course-materials/main/data/" + filename

flights = pd.read_csv(path)
print("予定便:", len(flights))
flights.head()
```
