## データを読み込む

次は，第 1 章と同じ運航記録を読めるか確かめます。次のセルは，CSV をノートと同じフォルダー，教材の `data/`，一つ上のフォルダーの `data/` の順に探します。どこにもなければ，公開 URL を使います。この準備セルはそのまま実行してください。

```python
from pathlib import Path

filename = "nyc-flights-2024-01.csv"
path = Path(filename)
if not path.exists():
    path = Path("data") / filename
if not path.exists():
    path = Path("../data") / filename

if not path.is_file():
    path = "https://raw.githubusercontent.com/sohei-takagi/python-ds-course-materials/main/data/" + filename

flights = pd.read_csv(path)
print("予定便:", len(flights))
flights.head()
```
